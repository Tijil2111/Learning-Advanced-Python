### Numpy Indexing and Arrays

```python
arr = np.arange(0,11)
```

```python
arr
```

```python
arr[8]
```

```python
arr[1:5]
```

```python
arr_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
```

```python
arr_2d
```

```python
arr_2d[0,1]
```

```python
arr_2d[:2,1:]
```

```python
arr_2d = np.append(arr_2d,[50,55,60])
```

```python
arr_2d
```

```python
arr_2d.reshape(3,6)
```

```python
arr = np.arange(1,11)
```

```python
arr
```

```python
bool_arr = arr > 5
```

```python
bool_arr
```

```python
arr[bool_arr] # only returns elements where the boolean array is true, this is called conditional indexing
```

```python
arr_2d = np.arange(50).reshape(5,10)
```

```python
arr_2d[1:3,3:5] # format is arr_name[rows slice, cols slice]
```
