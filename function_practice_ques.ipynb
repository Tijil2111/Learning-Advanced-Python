```python
# lesser of two evens 

def less_of_two_evens(a,b):
    if a%2==0 and b%2==0:
        return min(a,b)
    else:
        return max(a,b)
```

```python
less_of_two_evens(2,4)
```

```python
less_of_two_evens(2,5)
```

```python
# animal crackers 
def animal_crackers(txt):
    word_list = txt.split()
    if word_list[0][0] == word_list[-1][0]:
        return True
    else:
        return False
```

```python
animal_crackers('Levelheaded Llama')
```

```python
animal_crackers('Crazy Kangaroo')
```

```python
#Old Macdonald

def old_macdonald(name):
    newStr = ''
    for i in range(len(name)):
        if i == 0 or i == 3:
            newStr += name[i].upper()
        else:
            newStr += name[i].lower()
    return newStr

```

```python
old_macdonald('macdonald')
```

```python
# master yoda 

def master_yoda(txt):
    words_list = txt.split()
    reverse_list = words_list[::-1]
    return " ".join(reverse_list)
```

```python
master_yoda('I am home')
```

```python
master_yoda('We are ready')
```

```python
# almost there 
def almost_there(n):
    if (n >=90 and n<= 110) or (n>=190 and n<=210):
        return True
    else:
        return False
```

```python
almost_there(104)
```

```python
almost_there(150)
```

```python
almost_there(209)
```

```python
# has 33

def has_33(nums):
    for i in range(len(nums)-1):
        if nums[i] == 3:
            if nums[i+1] == 3:
                return True
            else:
                return False
```

```python
has_33([1, 3, 3])
```

```python
has_33([1, 3, 1, 3])
```

```python
has_33([3, 1, 3])
```

```python
# paper doll
def paper_doll(txt):
    newStr = ""
    for i in txt:
        newStr += i*3
    return newStr
```

```python
paper_doll('Hello')
```

```python
# blacjack 

def blackjack(a,b,c):
    sum = a+b+c
    if a==11 or b==11 or c==11:
        sum -=10
    if sum <=21:
        return sum 
    else:
        return "BUST"
```

```python
blackjack(5,6,7)
```

```python
blackjack(9,9,9)
```

```python
blackjack(9,9,11)
```
