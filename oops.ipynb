```python
class User:
    def __init__(self,name,email,password):
        self.name = name
        self._email = email
        self.password = password

    def get_email(self):
        return self._email
    def set_email(self,newEmail):
        self._email = newEmail
```

```python
user1 = User("tijil","tijil2008@gmail.com","123456")
```

```python
user1.get_email()
```

```python
user1.set_email("hello@gmail.com")
```
